## Intial Data Observation
1. Number of unique customers are less than total transaction records indicating one customer has several transaction records
2. Around 21% of customer_ids are missing in 2009 - 2010 and 26% of customer_ids are missing in 2010 - 2011 transaction sheets
3. Negative quantities occur in the transaction data and appear to represent cancellations/returns, while negative unit prices also exist and require      separate investigation.
4. Value in Price field is unit price instead of whole transaction price
5. The number of observed customer IDs differs between the two sheets, but this difference alone cannot establish customer churn. Customer retention needs to be analyzed using customer activity over time and an explicit churn definition.
6. Approximately 1.94% of transaction-line records have cancellation-style invoice identifiers, while approximately 98.06% are normal invoice records.
7. negative quantity rows most(10205) of them are having invoice number starting with C while other (2121) rows are starting with normal invoice but with missing customer id
8. negative price rows are only 3 in 2009 - 2010 and all of them having invoice with starting A and stock code as B
9. one invoice have multiple rows indicating one invoice contain multiple product orders
10. customer purchase days total unique are 4383 with minimum 1 and maximum 267 and customer order total unique are 4383 with minimum 1 and maximum 270
11. The median purchase interval is approximately 24.7 days, while the mean is 51.7 days, indicating a strongly right-skewed distribution.
12. Approximately 75% of observed purchase intervals are below 62 days, suggesting that extended inactivity beyond this range may represent meaningful behavioral change.


## Proposed Churn Defination : 

1. Oberservation Window : 180 Days
2. Prediction Window : 90 Days
3. Churn Defination : A customer is considered behaviorally churned if they make no qualifying purchase during the 90-day prediction window.

### Rationale:
1. The customer-level median purchase interval is approximately 56 days, while the 75th percentile is approximately 118 days. A 180-day observation window therefore provides sufficient historical context to capture multiple purchasing cycles for many customers and allows us to engineer meaningful behavioral features such as recency, frequency, monetary value, purchase intervals, and product diversity.
2. A 90-day prediction window provides a sufficiently long future period to identify meaningful customer inactivity while remaining operationally useful for retention planning.
3. Prediction snapshots will only be created when the complete 90-day future period exists in the dataset. This prevents right-censored observations from being incorrectly labeled as churn.
